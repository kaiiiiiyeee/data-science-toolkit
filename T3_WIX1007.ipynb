```python
import numpy as np
```

1. Create a numpy array of size 2 x 2

```python
n1 = np.array([[1,2],[3,4]])
print(n1)
```

2. Create a random numpy array of size 3 x 4

```python
n2 = np.random.random((3,4))
print(n2)
```

3. Create a numpy array of ones and a numpy array of zeroes

```python
n3 = np.zeros((2,2))
n4 = np.ones((2,2))
print(n3)
print(n4)
```

4. Create a random numpy array of size 6 x 3 and transpose it

```python
n5 = np.random.random((6,3))
transposed_5 = n5.T
print(transposed_5)
```

5. Create two random numpy array of size n x p and p x m and then compute the dot product

```python
n = 3
p = 4
m = 2
n6 = np.random.random((n,p))
n7 = np.random.random((p,m))
dot_product = n6 @ n7
print(n6)
print(n7)
print(dot_product)
```
